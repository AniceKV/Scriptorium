```python
import matplotlib.pyplot as plt
import re

log = """
step 0: train loss 4.4609, val loss 4.4688
step 100: train loss 2.2911, val loss 2.3708
step 200: train loss 1.8878, val loss 2.0131
step 300: train loss 1.5861, val loss 1.7465
step 400: train loss 1.4425, val loss 1.6327
step 500: train loss 1.3456, val loss 1.5423
step 600: train loss 1.2911, val loss 1.5033
step 700: train loss 1.2412, val loss 1.4578
step 800: train loss 1.2007, val loss 1.4246
step 900: train loss 1.1737, val loss 1.4069
step 1000: train loss 1.1490, val loss 1.3817
step 1100: train loss 1.1289, val loss 1.3650
step 1200: train loss 1.1086, val loss 1.3480
step 1300: train loss 1.0924, val loss 1.3308
step 1400: train loss 1.0787, val loss 1.3251
step 1500: train loss 1.0664, val loss 1.3227
step 1600: train loss 1.0524, val loss 1.3083
step 1700: train loss 1.0385, val loss 1.2993
step 1800: train loss 1.0281, val loss 1.2891
step 1900: train loss 1.0194, val loss 1.2881
step 2000: train loss 1.0108, val loss 1.2806
step 2100: train loss 1.0067, val loss 1.2871
step 2200: train loss 0.9962, val loss 1.2765
step 2300: train loss 0.9906, val loss 1.2743
step 7500: train loss 0.7936, val loss 1.1993
"""

steps, train_losses, val_losses = [], [], []
for line in log.strip().split('\n'):
    match = re.search(r'step (\d+): train loss ([\d.]+), val loss ([\d.]+)', line)
    if match:
        steps.append(int(match.group(1)))
        train_losses.append(float(match.group(2)))
        val_losses.append(float(match.group(3)))

plt.figure(figsize=(10, 5))
plt.plot(steps, train_losses, label='Train Loss', color='steelblue', marker='o', markersize=3)
plt.plot(steps, val_losses, label='Val Loss', color='orange', marker='o', markersize=3)
plt.xlabel('Steps')
plt.ylabel('Loss')
plt.title('BibleGPT — Training Curve (19M param GPT, trained from scratch)')
plt.legend()
plt.grid(alpha=0.3)
plt.annotate('Early stopping\nval loss: 1.1993', xy=(7500, 1.1993),
             xytext=(5500, 1.35), arrowprops=dict(arrowstyle='->'), fontsize=9)
plt.tight_layout()
plt.savefig('loss_curve.png', dpi=150)
plt.show()
```
